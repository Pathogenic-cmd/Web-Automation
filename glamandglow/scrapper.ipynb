```python
import time
import undetected_chromedriver as uc
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

def scrape_product_links(url):
    """
    Scrapes product links page by page.
    Instead of clicking 'Next' automatically, 
    the script pauses so you can manually click 'Next' in the browser.
    """
    all_links = set()

    driver = uc.Chrome()
    driver.get(url)

    try:
        page = 1
        while True:
            print(f"\n🌍 On page {page}")

            # Wait for products to load
            WebDriverWait(driver, 20).until(
                EC.presence_of_all_elements_located((By.CSS_SELECTOR, 'div.product__title a'))
            )

            # Extract product links
            product_elements = driver.find_elements(By.CSS_SELECTOR, 'div.product__title a')
            for element in product_elements:
                link = element.get_attribute('href')
                if link:
                    all_links.add(link)

            print(f"✅ Found {len(product_elements)} links on this page. Total unique so far: {len(all_links)}")

            # Pause here and let user click "Next" manually
            input("👉 Click 'Next' in the browser, then press ENTER here to continue... ")

            page += 1

    except KeyboardInterrupt:
        print("🛑 Stopped by user.")
    finally:
        driver.quit()

    return all_links


if __name__ == "__main__":
    start_url = "https://glamandglowbeautyhub.shop/collections/all?page=1"
    links = scrape_product_links(start_url)

    print("\n--- All Unique Product Links ---")
    if links:
        for link in sorted(links):
            print(link)
        print(f"\n🎯 Total unique links found: {len(links)}")
    else:
        print("No links were found.")
```

```python
import undetected_chromedriver as uc
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

def get_product_images(url):
    """
    Extracts the highest resolution product image URLs from a product page.

    Args:
        url (str): Product page URL.

    Returns:
        list[str]: List of high-res image URLs.
    """
    driver = uc.Chrome()
    driver.get(url)

    image_urls = []

    try:
        # Wait for product images to load
        img_elements = WebDriverWait(driver, 15).until(
            EC.presence_of_all_elements_located(
                (By.CSS_SELECTOR, 'img.product-single__photo')
            )
        )

        for img in img_elements:
            srcset = img.get_attribute("srcset")
            if srcset:
                # Split by commas → each entry is "url size"
                parts = [s.strip() for s in srcset.split(",")]
                # Last one should be highest resolution
                highest = parts[-1].split()[0]
                if highest.startswith("//"):
                    highest = "https:" + highest
                image_urls.append(highest)
            else:
                # fallback to src
                src = img.get_attribute("src")
                if src:
                    if src.startswith("//"):
                        src = "https:" + src
                    image_urls.append(src)

    except Exception as e:
        print(f"❌ Could not extract images: {e}")
    finally:
        driver.quit()

    return list(set(image_urls))  # unique only


# --- Test it ---
if __name__ == "__main__":
    test_url = "https://glamandglowbeautyhub.shop/products/advanced-clinicals-retinol-body-cream"
    images = get_product_images(test_url)
    print("🖼 Product Images (highest resolution only):")
    for img in images:
        print(img)
```

```python
import os
import time
import logging
import re
import pandas as pd
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.common.exceptions import WebDriverException
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager
import warnings
import urllib3

# ---------------- SETUP ----------------
warnings.filterwarnings("ignore", category=UserWarning, module="urllib3")
urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s",
    handlers=[logging.StreamHandler()]
)

# ---------------- CONFIG ----------------
PAGE_LOAD_TIMEOUT = 50
SCROLL_PAUSE = 2

columns = [
    "Product ID", "Product Name", "Product Type", "Category", "Brand Name",
    "Product Line Name", "Ingredients", "Use Instructions", "Package Size",
    "Product Description", "Product Colour", "Country of Origin", "Date Added",
    "Barcode (EAN/UPC)", "Barcode Type (e.g., EAN-13, UPC-A)", "Batch Number",
    "SKU", "Benefits", "Product Image URL", "Hero Ingriedients Match",
    "Key Ingriedients", "Key Ingriedients2", "Key Ingriedients3",
    "Key Ingriedients4", "Key Ingriedients5", "Product Images",
    "Verification Status", "Verification Date",
    "Notes (For internal use: flags, manual checks, comments, etc.)", "Source"
]

# ---------------- OUTPUT PATHS ----------------
BASE_DIR = os.getcwd()
OUTPUT_DIR = os.path.join(BASE_DIR, "output")
os.makedirs(OUTPUT_DIR, exist_ok=True)
filename = os.path.join(OUTPUT_DIR, "scraped_data.csv")
no_image_file = os.path.join(OUTPUT_DIR, "no_images.csv")

# ---------------- RESUME PROGRESS ----------------
if os.path.exists(filename):
    df_existing = pd.read_csv(filename)
    d = {col: df_existing[col].tolist() for col in df_existing.columns}
    scraped_urls = set(df_existing["Source"])
    logging.info(f"Resuming scrape. {len(scraped_urls)} URLs already scraped.")
else:
    d = {col: [] for col in columns}
    scraped_urls = set()

if os.path.exists(no_image_file):
    df_noimg = pd.read_csv(no_image_file)
    no_image_urls = set(df_noimg["URL"])
else:
    no_image_urls = set()

# ---------------- SAVE PROGRESS ----------------
def save_progress():
    df = pd.DataFrame(d)
    df.to_csv(filename, index=False)
    pd.DataFrame({"URL": list(no_image_urls)}).to_csv(no_image_file, index=False)
    logging.info(
        f"💾 Saved progress successfully! Scraped data: {len(d['Source'])} products, "
        f"No-image URLs: {len(no_image_urls)}"
    )

# ---------------- SELENIUM HELPERS ----------------
def safe_get(driver, url, retries=3):
    for attempt in range(retries):
        try:
            driver.get(url)
            time.sleep(SCROLL_PAUSE)
            return True
        except WebDriverException:
            logging.warning(f"⚠️ Attempt {attempt+1} failed loading {url}, retrying …")
            time.sleep(SCROLL_PAUSE)
    logging.error(f"❌ Could not load {url}")
    return False

def normalize_shopify_image(url: str) -> str:
    if not url:
        return ""
    if url.startswith("//"):
        url = "https:" + url
    return re.sub(r'_(small|medium|large|grande|compact|x\d+)+(\.[a-z]{3,4})(\?.*)?$', r'\2', url)

def product_type():
    return "Beauty"

# ---------------- COMBINED PRODUCT SCRAPER ----------------
def scrape_product_page(driver, url):
    """
    Extracts product name, sizes, description, and images from a product page.
    """
    result = {"name": "", "sizes": [], "description": "", "images": []}

    logging.info(f"   → Loading product page: {url}")
    if not safe_get(driver, url):
        logging.error(f"❌ Skipping {url} due to load failure")
        return result

    # Product Name
    try:
        element = WebDriverWait(driver, 15).until(
            EC.presence_of_element_located((By.CSS_SELECTOR, 'h1.product-single__title[itemprop="name"]'))
        )
        result["name"] = element.text.strip()
        logging.info(f"   → Product Name: {result['name'][:60]}")
    except Exception as e:
        logging.warning(f"⚠️ Could not extract product name: {e}")

    # Package Sizes
    try:
        select_element = WebDriverWait(driver, 5).until(
            EC.presence_of_element_located((By.CSS_SELECTOR, 'select.single-option-selector[data-option="option1"]'))
        )
        options = select_element.find_elements(By.TAG_NAME, "option")
        result["sizes"] = [opt.get_attribute("value").strip() for opt in options if opt.get_attribute("value")]
        logging.info(f"   → Package Sizes: {result['sizes']}")
    except Exception:
        logging.info("   → No package sizes found")

    # Product Description
    try:
        desc_element = WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.CSS_SELECTOR, 'div.product-single__description.rte'))
        )
        result["description"] = desc_element.text.strip()
        logging.info(f"   → Description length: {len(result['description'])} chars")
    except Exception:
        logging.info("   → No description found")

    # Product Images
    try:
        img_elements = WebDriverWait(driver, 10).until(
            EC.presence_of_all_elements_located((By.CSS_SELECTOR, 'img.product-single__photo'))
        )
        images = []
        for img in img_elements:
            srcset = img.get_attribute("srcset")
            if srcset:
                parts = [s.strip() for s in srcset.split(",")]
                highest = parts[-1].split()[0]
                if highest.startswith("//"):
                    highest = "https:" + highest
                images.append(highest)
            else:
                src = img.get_attribute("src")
                if src:
                    if src.startswith("//"):
                        src = "https:" + src
                    images.append(src)
        result["images"] = list(set(images))
        logging.info(f"   → Found {len(result['images'])} images")
    except Exception:
        logging.info("   → No images found")

    return result

from datetime import datetime

# ---------------- MAIN SCRAPER ----------------
options = Options()
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_argument("--start-maximized")
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)

BASE_URL = "https://glamandglowbeautyhub.shop"
# links_df = pd.read_csv("redo.csv")
# csv_links = links_df["Source"].dropna().tolist()
product_links = [link if link.startswith("http") else BASE_URL + link for link in links]

for idx, url in enumerate(product_links, 1):
    if url in scraped_urls:
        logging.info(f"[{idx}/{len(product_links)}] Skipping already scraped {url}")
        continue

    logging.info(f"[{idx}/{len(product_links)}] Scraping {url}")
    data = scrape_product_page(driver, url)

    if not data["images"]:
        no_image_urls.add(url)
        logging.info("⚠️ No images found, added to no_image_urls")

    today_str = datetime.today().strftime("%Y-%m-%d")  # <-- Today's date

    # Append data to global dict
    d["Product ID"].append("")
    d["Product Name"].append(data["name"])
    d["Product Type"].append("")
    d["Category"].append("")
    d["Brand Name"].append("")
    d["Product Line Name"].append("")
    d["Ingredients"].append("")
    d["Use Instructions"].append("")
    d["Package Size"].append(", ".join(data["sizes"]))
    d["Product Description"].append(data["description"])
    d["Product Colour"].append("")
    d["Country of Origin"].append("")
    d["Date Added"].append(today_str)
    d["Barcode (EAN/UPC)"].append("")
    d["Barcode Type (e.g., EAN-13, UPC-A)"].append("")
    d["Batch Number"].append("")
    d["SKU"].append("")
    d["Benefits"].append("")
    d["Product Image URL"].append(", ".join(data["images"]))
    d["Hero Ingriedients Match"].append("")
    d["Product Images"].append("")
    d["Verification Status"].append("")
    d["Verification Date"].append("")
    d["Notes (For internal use: flags, manual checks, comments, etc.)"].append("")
    d["Source"].append(url)
    for col in ["Key Ingriedients", "Key Ingriedients2", "Key Ingriedients3",
                "Key Ingriedients4", "Key Ingriedients5"]:
        d[col].append("")

    scraped_urls.add(url)
    save_progress()

driver.quit()
logging.info("🎉 Scraping complete")
```
